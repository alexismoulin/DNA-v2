```python
from Bio.Blast import NCBIWWW, NCBIXML
from Bio.Seq import Seq
```

```python
dna_sequence = "TGGGCCTCATATTTATCCTATATACCATGTTCGTATGGTGGCGCGATGTTCTACGTGAATCCACGTTCGAAGGACATCATACCAAAGTCGTACATTAGGACCTCGATATGGTTTTATTCTGTTTATCGTATCGGAGGTTATGTTCTTTTTTGCTCTTTTTCGGGCTTCTTCTCATTCTTCTTTGGCACCTACGGTAGAG"
```

```python
result_handle = NCBIWWW.qblast(program="blastn", database="nt", sequence=dna_sequence)
```

```python
blast_record = NCBIXML.read(handle=result_handle)
```

```python
len(blast_record.alignments)
```

```python
E_VALUE_THRESH: float = 0.01
```

```python
for alignment in blast_record.alignments:
    for hsp in alignment.hsps:
        if (hsp.expect < E_VALUE_THRESH and (
            "Nicotiana tabacum" in alignment.title
            or "Viburnum dentatum" in alignment.title
            or "Rhazya stricta" in alignment.title
            or "Sambucus canadensis" in alignment.title
        )):
            print("**** Alignment ****")
            print(f"sequence: {alignment.title}")
            print(f"length: {alignment.length}")
            print(f"E-value: {hsp.expect}")
            print(hsp.query)
            print(hsp.match)
            print(hsp.sbjct)
```

```python
my_seq = Seq(data=dna_sequence)
```
